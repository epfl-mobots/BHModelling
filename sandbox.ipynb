## Just some sandbox to try plotly

```python
# Imports
from libbh import *
```

### Configuration

```python
# Hive configuration
hive_depth = 10
hive_width = 9
hive_height = 265/460*hive_depth
hive_alpha = 0.2

# Frames configuration
frames_alpha = 0.6
frames_pos = [1,2,3,4] # Positions of the frames to include in the hive. Pos1 is on the right and Pos10 on the left.

# ABCs configuration
abcs_pos = [5,6] # Positions of the ABC robots to include in the hive. Pos1 is on the right and Pos10 on the left.
abcs_names = ["abc01", "abc02"] # Names of the ABC robots

# Partition configuration
partition_pos = 7

```

### Create the hive

```python
hive = BoxHive(hive_width, hive_height, hive_depth, hive_alpha)
hive.draw_plotly(show_legend=True)
hive.add_partition(partition_pos, show_legend=True)
```

### Add the frames

```python
for pos in frames_pos:
    hive.add_frame(pos, frames_alpha, show_legend=True)

# frames = []
# for pos in frames_pos:
#     x = hive.frame_positioning[pos]
#     _frame = Frame(
#         x=x,
#         height=hive.frame_height,
#         depth=hive.frame_depth,
#         opacity=frames_alpha,
#     )
#     _frame.draw_plotly(hive.height, hive.depth,"gold","frame", show_legend=True)
#     frames.append(_frame)
```

### Add the ABCs

```python
abcs = []
for pos in abcs_pos:
    x = hive.frame_positioning[pos]
    _abc = ABC(
        x=x,
        height=hive.frame_height,
        depth=hive.frame_depth,
        opacity=frames_alpha
    )
    _abc.draw_plotly(hive.height,hive.depth,"red", show_legend=True)
```

### Show result and annotate with axis

```python
textures = hive.getAllTextures()
# for _frame in frames:
#     textures +=_frame.get_textures()

fig = go.Figure(data=textures)

# Order the legend alphabetically


fig.update_layout(
    scene=dict(
        xaxis=dict(visible=False),
        yaxis=dict(visible=False),
        zaxis=dict(visible=False),
    ),
    title="3D Hive with Frames",
)

# Show the origin and axes with labels
fig.add_trace(go.Scatter3d(
    x=[0], y=[0], z=[0],
    mode="markers+text",
    marker=dict(size=5, color="red"),
    name="Origin",
    text=["Origin"],
    textposition="bottom right"
))
fig.add_trace(go.Scatter3d(
    x=[10], y=[0], z=[0],
    mode="markers+text",
    marker=dict(size=5, color="blue"),
    name="X Axis",
    text=["X"],
    textposition="top right"
))
fig.add_trace(go.Scatter3d(
    x=[0], y=[10], z=[0],
    mode="markers+text",
    marker=dict(size=5, color="green"),
    name="Y Axis",
    text=["Y"],
    textposition="top right"
))
fig.add_trace(go.Scatter3d(
    x=[0], y=[0], z=[10],
    mode="markers+text",
    marker=dict(size=5, color="yellow"),
    name="Z Axis",
    text=["Z"],
    textposition="top right"
))

fig.show()
```
